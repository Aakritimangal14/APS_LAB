# MATRIX CHAIN MULTIPLICATION Dynamic Programming with table filling.
Problem Description:
Find minimum number of scalar multiplications needed to multiply a chain of matrices..

#Theory
Cost of multiplying two matrices:
If A is p×q and B is q×r, cost = p × q × r

Dynamic Programming idea:
Break problem into smaller subproblems.
Store results in a table to avoid recomputation.

m[i][j] = minimum cost to multiply Ai to Aj

## Algorithm
1. Take array p[] of size n
2. Create table m[n][n]
3. Set m[i][i] = 0
4. Fill table diagonally
5. For each chain length L:
    Compute m[i][j]
    Try all possible k between i and j
6. Choose minimum cost
7. Return m[1][n-1]

```python
import sys

# Function to print optimal parenthesization
def print_optimal_parenthesis(s, i, j):
    if i == j:
        return f"A{i}"
    return "(" + \
           print_optimal_parenthesis(s, i, s[i][j]) + \
           print_optimal_parenthesis(s, s[i][j] + 1, j) + \
           ")"


def matrix_chain_order(p):
    n = len(p)

    m = [[0] * n for _ in range(n)]
    s = [[0] * n for _ in range(n)]

    for L in range(2, n):
        for i in range(1, n - L + 1):
            j = i + L - 1
            m[i][j] = sys.maxsize

            for k in range(i, j):
                cost = (m[i][k] +
                        m[k + 1][j] +
                        p[i - 1] * p[k] * p[j])

                if cost < m[i][j]:
                    m[i][j] = cost
                    s[i][j] = k

    return m, s


# Naive left-to-right multiplication
def naive_cost(p):
    cost = 0
    rows = p[0]
    cols = p[1]

    for i in range(1, len(p) - 1):
        cost += rows * cols * p[i + 1]
        cols = p[i + 1]

    return cost


# -------- INPUT --------
n = int(input("Enter number of dimensions: "))
print("Enter the dimensions array (space-separated):")
p = list(map(int, input().split()))

if len(p) != n:
    print("Error: number of elements must match n")
    exit()

# -------- PROCESS --------
m, s = matrix_chain_order(p)

min_cost = m[1][n - 1]
optimal_parenthesis = print_optimal_parenthesis(s, 1, n - 1)

naive = naive_cost(p)
saved = naive - min_cost

# -------- OUTPUT --------
print("\nMinimum multiplication cost:", min_cost)

print("\nOptimal Parenthesization:")
print(optimal_parenthesis)

print("\nNaive (left-to-right) cost:", naive)


print("Multiplications saved:", saved)
```
