# Minimum Cost Spanning Tree
Problem Description:
A spanning tree of a graph can be defined as a graph with minimal set of edges that connect all vertices. A minimum spanning tree of a graph is a spanning tree of the graph with least weight (where the weight is computed by adding the weights of all the edges in the spanning tree). In general, a graph can have multiple minimum spanning trees. The problem is to find a minimum spanning tree of a graph.

This implementation uses a dictionary to represent the weighted graph, where the vertices are keys, and the values are dictionaries containing their adjacent vertices and their weights.

Please note that this implementation assumes that the input graph is connected and weighted.

First, let's create a simple graph to demonstrate Prim's algorithm:

```python
graph = {
    "A": {"B": 10, "C": 15, "D": 20},
    "B": {"A": 10, "C": 12},
    "C": {"A": 15, "B": 12, "D": 18},
    "D": {"A": 20, "C": 18},
}
```

Now, let's implement Prim's algorithm:

```python
def prim_algorithm(graph):
    min_tree = {}
    visited = set()
    min_tree[list(graph.keys())[0]] = []
    min_weight = 0

    while len(visited) < len(graph):
        min_edge = float('inf')
        next_vertex = None
        for vertex in min_tree:
            for adj_vertex, weight in graph[vertex].items():
                if adj_vertex not in visited:
                    if weight < min_edge:
                        min_edge = weight
                        next_vertex = adj_vertex

        min_tree[next_vertex] = []
        visited.add(next_vertex)
        min_weight += min_edge

    return min_tree, min_weight
```

Now, let's run the Prim's algorithm on the example graph:
Keep in mind that this example uses a simple representation of a graph, but for more complex cases, you might want to use a library such as NetworkX to represent and manipulate graphs.

```python
graph = {
    "A": {"B": 10, "C": 15, "D": 20},
    "B": {"A": 10, "C": 12},
    "C": {"A": 15, "B": 12, "D": 18},
    "D": {"A": 20, "C": 18},
}

def prim_algorithm(graph):
    min_tree = {}
    visited = set()
    # Start with an arbitrary vertex, e.g., the first one in the graph keys
    start_vertex = list(graph.keys())[0]
    min_tree[start_vertex] = []
    visited.add(start_vertex)

    total_min_weight = 0

    # Loop until all vertices are visited
    while len(visited) < len(graph):
        min_edge = float('inf')
        next_vertex_to_add = None
        edge_from_vertex = None

        # Iterate over vertices already in the MST
        for vertex in visited:
            # Iterate over adjacent edges of these vertices
            for adj_vertex, weight in graph[vertex].items():
                # If the adjacent vertex is not yet visited and the edge is smaller than current min_edge
                if adj_vertex not in visited:
                    if weight < min_edge:
                        min_edge = weight
                        next_vertex_to_add = adj_vertex
                        edge_from_vertex = vertex

        # Add the found minimum edge and vertex to the MST
        if next_vertex_to_add is not None:
            min_tree[edge_from_vertex].append((next_vertex_to_add, min_edge))
            if next_vertex_to_add not in min_tree:
                min_tree[next_vertex_to_add] = [] # Initialize list for the new vertex if it's not a key yet
            visited.add(next_vertex_to_add)
            total_min_weight += min_edge
        else:
            # This case should ideally not be reached if the graph is connected
            break

    return min_tree, total_min_weight

mst, min_weight = prim_algorithm(graph)
print("Minimum Spanning Tree: ", mst)
print("Minimum Weight: ", min_weight)
```

Use same apporach for Kruskal's algorithm

```python
class DisjointSet:
    def __init__(self, vertices):
        self.parent = {v: v for v in vertices}
        self.rank = {v: 0 for v in vertices}

    def find(self, item):
        if self.parent[item] == item:
            return item
        self.parent[item] = self.find(self.parent[item])
        return self.parent[item]

    def union(self, set1, set2):
        root1 = self.find(set1)
        root2 = self.find(set2)

        if root1 != root2:
            if self.rank[root1] < self.rank[root2]:
                self.parent[root1] = root2
            elif self.rank[root1] > self.rank[root2]:
                self.parent[root2] = root1
            else:
                self.parent[root2] = root1
                self.rank[root1] += 1
            return True
        return False

def kruskal_algorithm(graph):
    min_tree = []
    total_min_weight = 0
    edges = []

    # Extract all edges from the graph
    for u in graph:
        for v, weight in graph[u].items():
            # Avoid adding duplicate edges (e.g., A-B and B-A) by only adding one direction
            # This assumes an undirected graph where edge (u,v) is same as (v,u)
            if (weight, v, u) not in edges:
                edges.append((weight, u, v))

    # Sort edges by weight
    edges.sort()

    vertices = list(graph.keys())
    ds = DisjointSet(vertices)

    for weight, u, v in edges:
        if ds.find(u) != ds.find(v):
            ds.union(u, v)
            min_tree.append((u, v, weight))
            total_min_weight += weight

    return min_tree, total_min_weight

# Test with graph_sample
min_tree_kruskal, min_weight_kruskal = kruskal_algorithm(graph_sample)
print("Minimum Spanning Tree (Kruskal's) for sample graph: ", min_tree_kruskal)
print("Minimum Weight (Kruskal's) for sample graph: ", min_weight_kruskal)
```
